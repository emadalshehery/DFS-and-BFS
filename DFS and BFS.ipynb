```python
from collections import deque

def bfs(tree, start):
    visited = set()
    queue = deque([start])
    visited.add(start)
    
    while queue:
        vertex = queue.popleft()
        print(vertex, end=' ')
        
        for neighbor in tree[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': set(),
    'E': set(),
    'F': set()
}
bfs(tree, 'A')

print(" BFS ")

def dfs(tree, start, visited = None):
    if visited is None:
        visited = set()
    visited.add(start)
    print(start, end=' ')
    
    for neighbor in tree[start]:
        if neighbor not in visited:
            dfs(tree, neighbor, visited)
    


tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': set(),
    'E': set(),
    'F': set()
}
dfs(tree, 'A')

print(" DFS ")
```
